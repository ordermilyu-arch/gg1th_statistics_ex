```python
import numpy as np
from scipy import stats
import pandas as pd
```

```python
# 성적의 평균 구하기
x = [70, 60, 80, 60, 50]
print(np.sum(x)/len(x))
print(np.mean(x))
print(np.array(x).mean())
print(pd.Series(x).mean())
```

```python
x = np.array([60, 40])
print(round(len(x) / (1/60 + 1/40),1))
print(round(len(x) / np.sum(1/x),1))
print(round(stats.hmean(x),1))
```

```python
np.random.seed(0)
income = np.random.normal(3000000, 500000, 100) 
print(round(np.mean(income), 2))
income = np.append(income, 10**8) 
print(round(np.mean(income), 2)) 
```

```python
x = np.random.choice(['A', 'B', 'C'], 2000) # A, B, C라는 요소로 데이터 1000개 만들기
print(x[:20])
print(len(x))

#파이선 모듈
from collections import Counter
counter = Counter(x)
mode_value, mode_count = counter.most_common(1)[0]  # 최빈값과 빈도수 추출


print("최빈값:", mode_value)
print("최빈값의 빈도수:", mode_count)
```

```python
income = np.append(income, 10**8) 
print(round(np.mean(income), 2))

#파이선 모듈
from collections import Counter
counter = Counter(income)
mode_value, mode_count = counter.most_common(1)[0]  # 최빈값과 빈도수 추출


print("최빈값:", mode_value)
print("최빈값의 빈도수:", mode_count)
```

```python
import numpy as np
import pandas as pd

x1 = np.array([1,2,3,4,5])
x2 = x1*10

z1 = (x1 - x1.mean()) / x1.std()
z2 = (x2 -x2.mean()) / x2.std()

print(z1)
print(z2)
```

```python
#Min-max Scaling
z1 = (x1 - x1.min()) / (x1.max() - x1.min())
z2 = (x2 - x2.min()) / (x2.max() - x2.min())

print(z1)
print(z2)
```

```python
# sklearn을 이용한 스케일링을 위한 데이터 준비
X = pd.DataFrame({"X1":[1, 2, 3, 4, 5],
    "X2": [10, 20, 30, 40, 50]})


from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler() #인스턴스화
Z = scaler.fit_transform(X) #fit.transfom = ndaaray


pd.DataFrame(Z, columns=['X1','X2'])
```

```python
# 표준 정규분포에 따라서 데이터의 수준을 맞춤
# 표준 정규분포 : 
from sklearn.preprocessing import StandardScaler
ss_scaler = StandardScaler() # 인스턴스화
Z = ss_scaler.fit_transform(X) # fit_transform => ndarray
pd.DataFrame(Z, columns=['X1', 'X2'])
```
