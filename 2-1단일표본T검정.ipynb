- 표본 데이터 불러오기

- 표본 데이터가 정규성 검정하기
    - KSTEST

- 정규표본 이라면 따른다면 T검정, 
    - p-value < 0.05 : 귀무가설 기각, 대립가설이 유효할것'같다.' 
    - p-value >= 0.05 : 대립가설 기각, 귀무가설 채택


```python
# 표본 데이터 불러오기
with open("datas2/성인여성_키_데이터1.txt", "r") as f:
    data = f.read().split('\n')
    # print(data)
    # map(data_type, list_data)
    data = list(map(float, data))
    print(data)
    print(len(data))
```

```python
#평균값, 표준값 

from scipy import stats
import numpy as np
mean_value = np.mean(data)
std_dev = np.std(data, ddof=0)  # 모집단 표준편차(기본 설정)

mean_value, std_dev
```

# 가설설정
- 귀무가설 : 표본데이터의 평균은 163과 같다.
- 대립가설 : 표본데이터의 평균은 163이 다르다.

## 정규성 검정의 가설
- 귀무가설 : 수집 표본의 데이터는 정규성을 따른다. (정규성의 기본값?)
- 대립가설 : 수집한 표본의 데이터는 정규성을 따르지 않다.

# 표본 데이터 정규성 검정하기
    - 수집한 데이터가 정규분포를 따르는지 검증한다.

```python
#T-test 실행해서 정규성 검정

result = stats.kstest(
    data,
    stats.norm.cdf,
    args=(mean_value, std_dev)
)

print(result)


# statistic, p_value = stats.shapiro(data)

# print(statistic, p_value)
```

```python
#검정하기

print(stats.ttest_1samp(data, 163)) 
```

# p-value 가 0.05보다 작음으로 귀무가설을 기각하고 대립가설을 참으로 한다.
# 즉 163과 다르다 라는 결론

독립표본 t 검정


```python
import pandas as pd
import numpy as np
#라이브러리 import
```

```python
#파일 불러와서 인스턴스화 
df1 = pd.read_csv("datas2/반별_점수_type1.csv", encoding = "euc-kr")
# df1 = pd.read_csv("datas2/반별_점수_type1.csv")
df1.head()
```

```python
# a반 b반 분리

groupA=df1['점수'].loc[df1['반']=='A'].values
groupB=df1['점수'].loc[df1['반']=='B'].values

print(groupA)
print(groupB)
```

```python
# 각 반의 점수 평균
import numpy as np

# A반의 평균과 표준 편차
mean_A = np.mean(groupA)
std_A = np.std(groupA, ddof=1)  # 표본 표준 편차 (ddof=1)

# B반의 평균과 표준 편차
mean_B = np.mean(groupB)
std_B = np.std(groupB, ddof=1)  # 표본 표준 편차 (ddof=1)

# 결과 출력
print(f"A반 평균: {mean_A:.2f}, 표준 편차: {std_A:.2f}")
print(f"B반 평균: {mean_B:.2f}, 표준 편차: {std_B:.2f}")
```

```python
# 정리한 데이터의 정규성검정 (pvalue 가 0.05이상인가.)
print(stats.kstest(
  groupA, 
  stats.norm.cdf, 
  args=(np.mean(groupA), np.std(groupA, ddof=0))))

print(stats.kstest(
  groupB, 
  stats.norm.cdf, 
  args=(np.mean(groupB), np.std(groupB, ddof=0))))
```

정규성 검정
-정규성 검전에서 가설수립
    -대립가설 설정하기 - 두 그룹의 표본데이터는 정규분포을 따른다.
    -귀무가설 설정하기 - 두 그룹의 표본데이터는 정규분포를 따르지않는다.

[해석] pvaplue >= 0  귀마가설을 기각할 필요가없다.
즉 정규성을 따른다고 할수있다.



##등분산성 검정 (개별 확인후 같이 봤을때도 괜찮은지 보기)
- 가설 수립
    - 두 그룹은 분산이 유사하다(등분산성)
    - 두 그룹은 분산이 유사하지않다 (등)

```python
stats.levene(groupA, groupB) 
```

[해석]
p-value >= 0.005 두 그룹은 등분산성을 따른다.
p-value < 0.005 두 그룹은 등분산성을 따르지 않는다.

```python
print(stats.ttest_ind(groupA, groupB, equal_var = True)) 
```
