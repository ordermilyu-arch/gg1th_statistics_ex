# 필요기능 import

```python
import pandas as pd
df = pd.read_csv("datas2/다이어트약_효과검증.csv", encoding='utf-8')
df.columns
# A그룹
before = df['다이어트전(kg)']
# B그룹
after = df['다이어트후(kg)']


```

# 정규성 검증
    - p value  >0.05 면

```python
from scipy import stats

stats.shapiro(after - before) # Shapiro-Wilk 검정

#확인결과 pvalue 가 0.28로 정규성이 검증됨
```

# 쌍체 표본 t 검정
    - 귀무가설 = 다이어트 전과 후의 차이가 유의미하지않다 동일하다.      pvalue < 0.05
    - 대립가설 = 다이어트 전과 후의 차이가 유의미하다 . 다르다          pvalue >= 0.05

```python
stats.ttest_rel(before, after)
#결과값 p-value 5.37 즉 0.05보다 크기때문에 대립가설이 맞다
```

```python
import numpy as np
np.mean(before), np.mean(after)

np.mean(before) - np.mean(after)

# 전후 차이 계산(개인별 체중 차이)
diff = before - after

# 체중 차이의 평균 계산(평균 체중 감소량)
mean_diff = np.mean(diff)

# 체중 차이의 표준편차(감소량의 표준편차)
std_diff = np.std(diff, ddof=1)

# 체중 차이의 표준오차(표본에서 구한 평균 체중 감소량의 불확실성을 나타냄)
standard_error = std_diff / np.sqrt(len(diff))

# t-통계량 계산(평균 체중 감소량이 표준오차보다 몇 배 큰지를 나타냄)
t_statistic = mean_diff / standard_error
t_statistic
# 계산 결과 출력
print(mean_diff)       # 평균 체중 감소량
print(std_diff)        # 체중 감소량의 표준편차
print(standard_error)  # 평균 체중 감소량의 표준오차
print(t_statistic)     # t 통계량
```
